```python
import numpy as np
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
def normalize_cav(cav):
    return cav / (np.linalg.norm(cav) + 1e-12)


def compute_cav(model, pos_words, neg_words):
    pos_emb = model.encode(pos_words, convert_to_numpy=True)
    neg_emb = model.encode(neg_words, convert_to_numpy=True)
    cav = pos_emb.mean(axis=0) - neg_emb.mean(axis=0)
    return cav


def get_similarity_score(model, cav, words):
    word_emb = model.encode(words, convert_to_numpy=True)
    scores = np.dot(word_emb, cav)
    return scores
```

```python
pos = ["dogs", "cats", "cars", "houses", "trees", "bottles", "computers", "chairs"]
neg = ["dog", "cat", "car", "house", "tree", "bottle", "computer", "chair"]

cav = compute_cav(model, pos, neg)
cav_norm = normalize_cav(cav)
```

```python
# load text from a file
texts = None
with open("english-5k.csv", encoding="utf-8") as f:
    texts = [line.strip() for line in f]
```

```python
scores = get_similarity_score(model, cav_norm, texts)
```

```python
paired = list(zip(texts, scores))
paired.sort(key=lambda x: x[1], reverse=True)
```

```python
for text, score in paired:
    print(f"{text:<20}{score:.6f}")
```
